```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
  return(1/(1+np.exp(-z)))

X=np.array([1,0.5])

W1=np.array([[0.4,0.6],[-0.5,0.8]])
B1=np.array([0.1,-0.2])

W2=np.array([[0.7,-0.3]])
B2=np.array([0.05])

Z1=W1@X.T+B1
H1=sigmoid(Z1)

Z2=W2@H1.T+B2
H2=sigmoid(Z2)
```

```python
print(Z1,H1)
print(Z2,H2)
```

```python
plt.figure(figsize=(6, 3))

inputs = [(0, 1), (0, 0)]
hidden = [(1, 1), (1, 0)]
output = [(2, 0.5)]

for a in inputs:
    for b in hidden:
        plt.plot([a[0], b[0]], [a[1], b[1]], color="red", zorder=1)

for a in hidden:
    for b in output:
        plt.plot([a[0], b[0]], [a[1], b[1]], color="purple", zorder=1)

neurons = [
    (0, 1, "x1", X[0]),  (0, 0, "x2", X[1]),
    (1, 1, "h1", H1[0]), (1, 0, "h2", H1[1]),
    (2, 0.5, "out", H2[0]),
]

for x, y, name, value in neurons:
    plt.scatter(x, y, s=2000, color="orange", edgecolors="black")
    plt.text(x, y, f"{name}\n{value:.2f}", ha="center", va="center")

plt.xlim(-0.5,2.5)
plt.ylim(-0.5,1.5)

plt.axis("off")
plt.show()
```
